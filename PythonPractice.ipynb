# Encapsulation 

```python
class Account:
    def __init__(self,salary,expenses):
        self.salary=salary
        self.expenses=expenses
    def __str__(self):
        return ("Salary"+ str(self.salary)+ "Expenses"+str(self.expenses))
    
class Sales:
    def __init__(self,sales):
        self.sales=sales
    def __str__(self):
        return ("Sales"+str(self.sales))
    
class Purches:
    def __init__(self,purches):
        self.purches=purches
    def __str__(self):
        return("Purches"+str(self.purches))
    
class HumanResources:
    def __init__(self,hiring):
        self.hiring=hiring
    def __str__(self):
        return("Hiring"+str(self.hiring))

class Company:
    def __init__(self,account,sales,purches,hr):
        self.account=account
        self.sales=sales
        self.purches=purches
        self.hr=hr
    
    def __str__(self):
        return(str(self.account)+"\n"+str(self.sales)+"\n"+str(self.purches)+"\n"+str(self.hr))
    
```

```python
acc =Account(120000,5000)
sales = Sales(1000)
purches =Purches(2000)
hr = HumanResources("Employee")
```

```python
company =Company(acc,sales,purches,hr)
```

```python
print(company)
```

```python
expectations = [10, 30,6, 200]
cards = [5, 30,6, 200] 
#compare cards each values with expception each values if cards values are greater than expectations values then condition true and print greater values counts 


def max_satisfaction(expectations, cards):
  expectations.sort()
  cards.sort()
  exp_index=0
  card_index=0
  while exp_index < len(expectations) and card_index < len(cards):
     if cards[card_index] >= expectations[exp_index]:   
        exp_index+=1    
     card_index+=1
 
  return exp_index  

a=max_satisfaction(expectations, cards)
print(a)
```

```python
import pandas as pd
from datetime import datetime, timedelta

# Set start and end dates
start_date = datetime(2023, 1, 1)  # Adjust as needed
end_date = datetime.today().date()  # Covers more than a year

# Generate date range
date_list = pd.date_range(start=start_date, end=end_date, freq='D')

# Initialize list for week numbers
week_numbers = []
week_num = 1
extra_days = 6  # Number of extra days for Week 53

for i, date in enumerate(date_list):
    if i % 7 == 0:  # Every 7 days, increase the week number
        if week_num == 53:  # If Week 53 is reached, extend it by extra_days
            extra_days -= 1
            if extra_days < 0:
                week_num = 1  # Reset to 01 after Week 53
                extra_days = 6
        else:
            week_num += 1 if extra_days == 6 else 0  # Continue normally if not in extra days

    week_numbers.append(week_num)

# Create DataFrame
df = pd.DataFrame({"Date": date_list, "Week_Number": week_numbers})

# Print sample output
df # Print first 60 days for verification
```

```python
def my_decor(func):
    def wrapper(*args,**kwargs):
        print("before function runs")
        result=func(*args,**kwargs)
        print("after function runs")
        return result
    return wrapper  

@my_decor

def say_hello(a,b):
    return a+b
print(say_hello(1,1))
```
